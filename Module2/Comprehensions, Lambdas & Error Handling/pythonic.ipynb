```python
# 1. List Comprehension
numbers = [1, 2, 3, 4, 5]

squares = [num ** 2 for num in numbers]
print("List Comprehension:", squares)


# 2. Dictionary Comprehension
square_dict = {num: num ** 2 for num in numbers}
print("\nDictionary Comprehension:", square_dict)

# 3. Set Comprehension
words = ["apple", "banana", "apple", "orange", "banana"]

unique_lengths = {len(word) for word in words}
print("\nSet Comprehension:", unique_lengths)

# 4. Generator Expression
cube_generator = (num ** 3 for num in numbers)

print("\nGenerator Expression:")
for cube in cube_generator:
    print(cube)

# 5. lambda
multiply = lambda a, b: a * b
print("\nLambda:", multiply(5, 6))

# 6. map()
mapped = list(map(lambda x: x * 2, numbers))
print("\nmap():", mapped)

# 7. filter()
evens = list(filter(lambda x: x % 2 == 0, numbers))
print("\nfilter():", evens)

# 8. sorted(key=...)
students = [
    ("Ali", 75),
    ("Sara", 92),
    ("Ahmed", 81),
]

sorted_students = sorted(students, key=lambda student: student[1])

print("\nsorted(key=...):")
print(sorted_students)

# 9. enumerate()
print("\nenumerate():")
for index, value in enumerate(words, start=1):
    print(index, value)

# 10. zip()
names = ["Ali", "Sara", "Ahmed"]
marks = [75, 92, 81]

print("\nzip():")
for name, mark in zip(names, marks):
    print(f"{name}: {mark}")

# 11. try / except / else / finally
print("\nException Handling")

try:
    number = int(input("Enter an integer: "))
except ValueError:
    print("Invalid input! Please enter a valid integer.")
else:
    print("You entered:", number)
finally:
    print("Execution completed.")

# 12. Raising Exceptions
def divide(a, b):
    if b == 0:
        raise ValueError("Denominator cannot be zero.")
    return a / b


try:
    print("\nDivision:", divide(10, 2))
except ValueError as error:
    print(error)

# 13. Custom Exception
class InvalidAgeError(Exception):
    """Raised when age is outside the valid range."""
    pass


def validate_age(age):
    if age < 18:
        raise InvalidAgeError("Age must be at least 18.")
    print("Age is valid.")


try:
    validate_age(16)
except InvalidAgeError as error:
    print("\nCustom Exception:", error)


# 14. Robust Input Validation
def get_positive_integer():
    while True:
        try:
            value = int(input("\nEnter a positive integer: "))

            if value <= 0:
                raise ValueError("Number must be greater than zero.")

            return value

        except ValueError as error:
            print(error)


positive_number = get_positive_integer()
print("Valid number entered:", positive_number)
```

```python
#matrix transpose
mat=[[1,0,1],[2,0,3],[3,1,3]]
print("Transpose:")
for i in range(len(mat[0])):
    for j in range(len(mat)):
        print(mat[j][i], end=" ")
    print()

#prime sieve
print("Prime sieve:")
num=int(input("Enter a number : "))
for i in range(2,num+1):
    for j in range(2,i):
        if i%j==0:
            break
    else:
        print(i)

#Star pattern set:
print("Star pattern set:")
#Right angled:
n = 5
for i in range(1, n + 1):
    print("*" * i)
#Inverted right angle:
n = 5
for i in range(n, 0, -1):
    print("*" * i)

#Pyramid:
n = 5
for i in range(n):
    print(" " * (n - i - 1) + "*" * (2 * i + 1))
#Inverted pyramid:
n = 5
for i in range(n, 0, -1):
    print(" " * (n - i) + "*" * (2 * i - 1))
#Diamond:
n = 5
for i in range(n):
    print(" " * (n - i - 1) + "*" * (2 * i + 1))
for i in range(n - 2, -1, -1):
    print(" " * (n - i - 1) + "*" * (2 * i + 1))

```

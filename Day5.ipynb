```python
import random

# FizzBuzz
def fizzBuzz():
    for i in range(1, 21):
        if i % 15 == 0:
            print("FizzBuzz")
        elif i % 3 == 0:
            print("Fizz")
        elif i % 5 == 0:
            print("Buzz")
        else:
            print(i)


# Number Guessing Game
def numberGuessingGame():
    print("Think of a number between 1 to 100")
    guess = int(input("Enter your guess : "))
    num = random.randint(1, 100)

    while guess != num:
        if guess > num:
            print("Too high")
        else:
            print("Too low")
        guess = int(input("Enter your guess again: "))

    print("Correct guess")


# Multiplication Table
def multiplicationTable():
    x = int(input("enter a number: "))
    for i in range(1, 10):
        print(x, "*", i, "=", x * i)


# Word Count
def wordCount():
    para = input("enter a paragraph : ")
    words = para.split()
    wordCount = {}

    for i in words:
        if i in wordCount:
            wordCount[i] += 1
        else:
            wordCount[i] = 1

    print(wordCount)


# Converters
def kmToMiles(km):
    return float(km) * 0.621371


def celsiusToFahrenheit(celsius):
    return celsius * 9 / 5 + 32


def kgToLbs(kg):
    return kg * 2.20462


def bmi(kg, height):
    return kg / (height * height)


def converters():
    weight = float(input("enter weight in kg : "))
    print(kgToLbs(weight), "pounds")

    height = float(input("enter height in meters : "))
    print("BMI is : ", bmi(weight, height))

    distance = float(input("Enter distance(km): "))
    print(kmToMiles(distance), "miles")

    temp = float(input("enter temperature in celsius: "))
    print(celsiusToFahrenheit(temp), "Fahrenheit")


# Contacts
def contacts():
    contactList = {
        "Wasil": 923111978797,
        "Ali": 923111978798,
        "Ahmed": 923111978799
    }

    print(contactList.values())

    name = input("enter name : ")

    if name in contactList:
        print("Phone Number : ", contactList[name])
    else:
        print("Contact not found")


# Calculator
def calculator():
    print("Hello World")

    num1 = int(input("enter number 1 : "))
    num2 = int(input("enter number 2 : "))
    operation = input("enter operation (+,-,*,/): ")

    if operation == "+":
        print("addition is : ", num1 + num2)
    elif operation == "-":
        print("subtraction is : ", num1 - num2)
    elif operation == "*":
        print("multiplication is : ", num1 * num2)
    elif operation == "/":
        print("division is : ", num1 / num2)
    else:
        print("invalid operation")


# Function Calls
fizzBuzz()
numberGuessingGame()
multiplicationTable()
wordCount()
converters()
contacts()
calculator()
```

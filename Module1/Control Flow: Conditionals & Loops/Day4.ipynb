```python
import random
# FizzBuzz
for i in range(1, 21):
    if i % 15 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
#Number guessing game
print("Think of a number between 1 to 100")
guess=int(input("Enter your guess : "))
num=random.randint(1,100)
while guess != num:
    if guess > num:
        print("Too high")
    else:
        print("Too low")
    guess=int(input("Enter your guess again: "))
print("Correct guess")


#Multiplication Table
x=int(input("enter a number: "))
for i in range(1,10):
    print(x,"*",i,"=",x*i)
```

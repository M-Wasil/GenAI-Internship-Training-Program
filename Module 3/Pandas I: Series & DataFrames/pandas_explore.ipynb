# Pandas I — Series & DataFrames

**Objective:** Load, inspect, select, filter and analyze tabular data.

```python
import pandas as pd
import numpy as np
```

## 1. Create/load a sales dataset

```python
import pandas as pd
import numpy as np

np.random.seed(42)

raw_sales = pd.DataFrame({
    "order_id": range(1001, 1016),
    "date": [
        "2026-01-03", "2026-01-05", "2026-01-07", "2026-01-08", "2026-01-12",
        "2026-01-15", "2026-01-18", "2026-01-20", "2026-01-22", "2026-01-25",
        "2026-01-27", "2026-01-29", "2026-02-02", "2026-02-04", "2026-02-06"
    ],
    "region": ["Karachi", "Lahore", "Islamabad", "Karachi", "Lahore",
               "Karachi", "Islamabad", "Karachi", "Lahore", "Islamabad",
               "Karachi", "Lahore", "Karachi", "Islamabad", "Lahore"],
    "product": ["Laptop", "Phone", "Tablet", "Headphones", "Laptop",
                "Phone", "Tablet", "Laptop", "Headphones", "Phone",
                "Tablet", "Laptop", "Phone", "Headphones", "Laptop"],
    "quantity": [2, 3, 1, 5, 2, 4, 2, 1, 6, 3, 2, 1, 5, 4, 2],
    "unit_price": [120000, 65000, 45000, 8000, 120000, 65000, 45000, 120000,
                   8000, 65000, 45000, 120000, 65000, 8000, 120000],
    "customer_rating": [4.5, 4.0, np.nan, 4.2, 4.8, 3.9, 4.1, np.nan,
                        4.4, 3.8, 4.6, 4.7, 4.0, 4.3, np.nan]
})
raw_sales["revenue"] = raw_sales["quantity"] * raw_sales["unit_price"]
raw_sales.head()
```

## 2. Series and DataFrame basics

```python
sales = raw_sales.copy()

region_series = sales["region"]
print("Series type:", type(region_series))
print("DataFrame shape:", sales.shape)
print("Columns:", sales.columns.tolist())
print("Index:", sales.index)
```

## 3. Inspecting the data

```python
print(sales.head())
print("\nInfo:")
sales.info()
print("\nDescribe:")
display(sales.describe(include="all"))
```

## 4. loc and iloc

```python
print("loc rows 0:3:")
display(sales.loc[0:3, ["order_id", "product", "revenue"]])

print("iloc rows 0:3:")
display(sales.iloc[0:3, 0:4])
```

## 5. Filtering

```python
high_value = sales[sales["revenue"] > 100000]
karachi = sales[sales["region"] == "Karachi"]
laptops = sales[sales["product"] == "Laptop"]

display(high_value)
display(karachi)
display(laptops)
```

## 6. Adding columns

```python
sales["discount"] = np.where(sales["quantity"] >= 4, 0.10, 0.05)
sales["net_revenue"] = sales["revenue"] * (1 - sales["discount"])

display(sales[["product", "quantity", "revenue", "discount", "net_revenue"]].head())
```

## 7. Sorting and indexing

```python
print("Highest revenue:")
display(sales.sort_values("revenue", ascending=False).head())

print("Highest quantity:")
display(sales.sort_values("quantity", ascending=False).head())
```

## 8. Ten insights

```python
insights = [
    f"1. Dataset contains {len(sales)} orders.",
    f"2. Dataset has {sales['product'].nunique()} unique products.",
    f"3. Dataset covers {sales['region'].nunique()} regions.",
    f"4. Total revenue is Rs. {sales['revenue'].sum():,.0f}.",
    f"5. Average order revenue is Rs. {sales['revenue'].mean():,.0f}.",
    f"6. Highest single-order revenue is Rs. {sales['revenue'].max():,.0f}.",
    f"7. Total quantity sold is {sales['quantity'].sum()} units.",
    f"8. Most frequently sold product is {sales['product'].value_counts().idxmax()}.",
    f"9. Region with the most orders is {sales['region'].value_counts().idxmax()}.",
    f"10. Average customer rating is {sales['customer_rating'].mean():.2f}."
]
print("\n".join(insights))
```

Pandas provides high-level tabular data structures and expressive indexing, filtering and inspection operations that form the foundation of practical data analysis.

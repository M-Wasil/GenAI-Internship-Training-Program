# NumPy Fundamentals Lab

**Objective:** Practice fast numerical computing with NumPy and compare vectorized operations with pure Python.

```python
import numpy as np
import time

print("NumPy version:", np.__version__)
```

## 1. ndarray vs Python lists

```python
py_list = [1, 2, 3, 4, 5]
arr = np.array([1, 2, 3, 4, 5], dtype=np.int64)

print("List:", py_list)
print("Array:", arr)
print("dtype:", arr.dtype)
print("shape:", arr.shape)
print("ndim:", arr.ndim)
print("size:", arr.size)
```

## 2. Creating arrays

```python
zeros = np.zeros((2, 3))
ones = np.ones((2, 3))
range_arr = np.arange(0, 10, 2)
lin = np.linspace(0, 1, 6)
identity = np.eye(3)

print("zeros:\n", zeros)
print("ones:\n", ones)
print("arange:", range_arr)
print("linspace:", lin)
print("identity:\n", identity)
```

## 3. Indexing, slicing and boolean masks

```python
a = np.arange(1, 13).reshape(3, 4)

print(a)
print("First row:", a[0])
print("Element [1,2]:", a[1, 2])
print("First two columns:\n", a[:, :2])

mask = a > 6
print("Values > 6:", a[mask])
```

## 4. Vectorization and broadcasting

```python
x = np.array([1, 2, 3, 4])
y = np.array([10, 20, 30, 40])

print("x + y:", x + y)
print("x * 2:", x * 2)
print("x squared:", x ** 2)

matrix = np.arange(12).reshape(3, 4)
bias = np.array([1, 2, 3, 4])
print("Broadcasting result:\n", matrix + bias)
```

## 5. Aggregations and axes

```python
data = np.array([[10, 20, 30], [40, 50, 60]])

print("sum:", data.sum())
print("mean:", data.mean())
print("std:", data.std())
print("column sums:", data.sum(axis=0))
print("row sums:", data.sum(axis=1))
print("column means:", data.mean(axis=0))
```

## 6. Reshaping and stacking

```python
a = np.arange(1, 7).reshape(2, 3)
b = np.arange(7, 13).reshape(2, 3)

print("Vertical stack:\n", np.vstack((a, b)))
print("Horizontal stack:\n", np.hstack((a, b)))
print("Reshaped:\n", np.arange(12).reshape(4, 3))
```

## 7. Random numbers and basic linear algebra

```python
rng = np.random.default_rng(42)

random_matrix = rng.normal(size=(3, 3))
random_vector = rng.normal(size=3)

print("Random matrix:\n", random_matrix)
print("Random vector:", random_vector)
print("Matrix-vector product:", random_matrix @ random_vector)
print("Transpose:\n", random_matrix.T)
print("Determinant:", np.linalg.det(random_matrix))
```

## 8. Vectorization speed comparison

The same computation is implemented with a Python loop and NumPy vectorization.

```python
n = 1_000_000
values = list(range(n))
np_values = np.arange(n)

start = time.perf_counter()
python_result = [x * x + 2 * x + 1 for x in values]
python_time = time.perf_counter() - start

start = time.perf_counter()
numpy_result = np_values * np_values + 2 * np_values + 1
numpy_time = time.perf_counter() - start

print(f"Pure Python: {python_time:.6f} seconds")
print(f"NumPy:       {numpy_time:.6f} seconds")
print(f"Speedup:     {python_time / numpy_time:.2f}x")
```

## Conclusion

NumPy arrays provide compact typed storage and vectorized operations. Broadcasting allows operations between compatible shapes without explicit Python loops, while aggregation and linear algebra routines provide efficient numerical computation.
